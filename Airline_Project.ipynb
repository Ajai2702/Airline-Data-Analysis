```python
total_passengers = len(df_clean)

satisfied = (
    df_clean["satisfaction"] == "satisfied"
).sum()

dissatisfied = total_passengers - satisfied

satisfaction_rate = (
    satisfied / total_passengers
) * 100

avg_age = df_clean["Age"].mean()

avg_distance = df_clean["Flight Distance"].mean()

avg_departure_delay = (
    df_clean["Departure Delay in Minutes"].mean()
)

avg_arrival_delay = (
    df_clean["Arrival Delay in Minutes"].mean()
)

print("Total Passengers:", total_passengers)
print("Satisfied Passengers:", satisfied)
print("Neutral/Dissatisfied:", dissatisfied)
print("Satisfaction Rate:", round(satisfaction_rate, 2), "%")
print("Average Age:", round(avg_age, 2))
print("Average Flight Distance:", round(avg_distance, 2))
print("Average Departure Delay:", round(avg_departure_delay, 2))
print("Average Arrival Delay:", round(avg_arrival_delay, 2))
```

```python
col = "Departure Delay in Minutes"

Q1 = df_clean[col].quantile(0.25)
Q3 = df_clean[col].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean[col] < lower_bound) |
    (df_clean[col] > upper_bound)
]

print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("Outlier count:", len(outliers))
```

```python
numeric_cols = [
    "Age",
    "Flight Distance",
    "Departure Delay in Minutes",
    "Arrival Delay in Minutes"
]

for col in numeric_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = (
        (df_clean[col] < lower) |
        (df_clean[col] > upper)
    ).sum()

    print(col, "outliers:", count)
```

```python
sat_percent = (
    df_clean["satisfaction"]
    .value_counts(normalize=True) * 100
)

print(sat_percent.round(2))
```

```python
class_rate = (
    df_clean.assign(
        is_satisfied=(
            df_clean["satisfaction"] == "satisfied"
        )
    )
    .groupby("Class", observed=True)["is_satisfied"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print(class_rate.round(2))
```

```python
customer_rate = (
    df_clean.assign(
        is_satisfied=(
            df_clean["satisfaction"] == "satisfied"
        )
    )
    .groupby("Customer Type", observed=True)["is_satisfied"]
    .mean()
    .mul(100)
)

print(customer_rate.round(2))
```

```python
travel_rate = (
    df_clean.assign(
        is_satisfied=(
            df_clean["satisfaction"] == "satisfied"
        )
    )
    .groupby("Type of Travel", observed=True)["is_satisfied"]
    .mean()
    .mul(100)
)

print(travel_rate.round(2))
```

```python
delay_summary = (
    df_clean.groupby("satisfaction", observed=True)[
        [
            "Departure Delay in Minutes",
            "Arrival Delay in Minutes"
        ]
    ]
    .agg(["mean", "median"])
)

print(delay_summary.round(2))
```

```python
df_clean["departure_delayed"] = np.where(
    df_clean["Departure Delay in Minutes"] > 0,
    "Delayed",
    "No Delay"
)

delay_sat = pd.crosstab(
    df_clean["departure_delayed"],
    df_clean["satisfaction"],
    normalize="index"
) * 100

print(delay_sat.round(2))
```

```python
delay_sat.plot(
    kind="bar",
    figsize=(8, 5)
)

plt.title("Satisfaction by Departure Delay Status")
plt.ylabel("Percentage")
plt.xticks(rotation=0)
plt.show()
```

```python
df_clean["delay_category"] = pd.cut(
    df_clean["Departure Delay in Minutes"],
    bins=[-1, 0, 15, 60, np.inf],
    labels=[
        "No Delay",
        "1-15 Minutes",
        "16-60 Minutes",
        "Over 60 Minutes"
    ]
)

delay_category_sat = pd.crosstab(
    df_clean["delay_category"],
    df_clean["satisfaction"],
    normalize="index"
) * 100

print(delay_category_sat.round(2))
```
